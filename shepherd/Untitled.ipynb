```python
import numpy as np

# Load the mean SPL matrix
spl = np.load("shepherd_data/patients/single_patient/one_patient_agg=mean_spl_matrix.npy")

print("Shape:", spl.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# If it's a single row, convert to 2D for heatmap
sns.heatmap(spl, cmap="viridis")
plt.xlabel("Node index")
plt.ylabel("Patient")
plt.title("Shortest Path Distances for Single Patient")
plt.show()
```

```python
import pandas as pd

spl_row = spl[0]  # only one patient
top_k = 20  # show 20 closest nodes
top_indices = np.argsort(spl_row)[:top_k]

df = pd.DataFrame({
    "node_index": top_indices,
    "spl": spl_row[top_indices]
})

df.plot(kind="bar", x="node_index", y="spl", color="skyblue", legend=False)
plt.ylabel("Shortest Path Length")
plt.title("Top 20 Closest Nodes for Single Patient")
plt.show()
```
